```python
import pandas as pd

# Load your JSON
df = pd.read_json("haunted_places.json")

# Clean the "Alcohol Abuse Per Capita" column
df["Alcohol Deaths Per Capita"] = pd.to_numeric(df["Alcohol deaths per capita"], errors="coerce")

# Drop rows with missing or NaN alcohol data
df_cleaned = df.dropna(subset=["Alcohol Deaths Per Capita", "state"])

# Group by state and take the mean (or sum) if there are multiple entries
df_by_state = df_cleaned.groupby("state", as_index=False)["Alcohol Deaths Per Capita"].mean()
```

```python
df_by_state.to_json("alcohol_abuse_by_state.json", orient="records", indent=2)
```
