```python
import pandas as pd

# Load the original JSON file
df = pd.read_json('aggregated_witnesses.json')

# Inspect the first few rows to check column names and data types
print(df.head())

# Check if the latitude and longitude are numeric and clean up if needed
df['latitude'] = pd.to_numeric(df['latitude'], errors='coerce')
df['longitude'] = pd.to_numeric(df['longitude'], errors='coerce')

# Ensure the 'Haunted Places Witness Count' is numeric and handle any non-numeric entries
df['Haunted Places Witness Count'] = pd.to_numeric(df['Haunted Places Witness Count'], errors='coerce')

# Drop rows with NaN values in latitude, longitude, or 'Haunted Places Witness Count'
df_cleaned = df.dropna(subset=['latitude', 'longitude', 'Haunted Places Witness Count'])

# Aggregate the data by 'latitude', 'longitude', and 'city'
agg_by_location = df_cleaned.groupby(["latitude", "longitude", "city"], as_index=False)["Haunted Places Witness Count"].sum()

# Rename the 'Haunted Places Witness Count' column to 'witness_count'
agg_by_location = agg_by_location.rename(columns={'Haunted Places Witness Count': 'witness_count'})

# Optional: Verify the cleaned and aggregated data
print(agg_by_location.head())

# Save the result to a new JSON file
agg_by_location.to_json('aggregated_witnesses_cleaned_aggregated.json', orient='records', lines=True)
```

```python
# Group by city, latitude, and longitude and sum the witness counts
agg_by_location = df.groupby(["latitude", "longitude", "city"], as_index=False)["Haunted Places Witness Count"].sum()

# Select only the relevant columns: city, latitude, longitude, and witness count
agg_by_location_cleaned = agg_by_location[["city", "latitude", "longitude", "Haunted Places Witness Count"]]

# Rename columns for consistency (optional)
agg_by_location_cleaned = agg_by_location_cleaned.rename(columns={
    "Haunted Places Witness Count": "witness_count"
})

# Save the cleaned DataFrame to a new JSON file
agg_by_location_cleaned.to_json("aggregated_witnesses_cleaned_aggregated.json", orient="records", lines=True)

# Show the cleaned data preview
agg_by_location_cleaned.head()
```

```python
# Assuming your cleaned DataFrame is called 'agg_by_location'
import json

# Convert DataFrame to dictionary (list of records)
json_data = agg_by_location.to_dict(orient="records")

# Convert the list of dictionaries to a JSON string
json_string = json.dumps(json_data)

# Save the JSON string to a file
with open('/content/aggregated_witnesses_cleaned.json', 'w') as f:
    f.write(json_string)

# Check that the file is saved correctly
import os
os.listdir('/content')
```
